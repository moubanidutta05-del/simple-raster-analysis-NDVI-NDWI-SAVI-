```python
Landsat7 = ee.ImageCollection("LANDSAT/LE07/C02/T1_L2")
L7_optical = Landsat7.select('SR_B1','SR_B2','SR_B3','SR_B4','SR_B5','SR_B7')
Taluk_L7 = L7_optical.filterBounds(Taluk).filterDate("2000-06-28","2000-12-15").filterMetadata('CLOUD_COVER','Less_than',10).mean().clip(Taluk)
vis_L72000543 = {'bands': ['SR_B5', 'SR_B4', 'SR_B3'], 'gamma': 3.16, 'min': 9229.1, 'max': 64385.9}
Taluk_L7 = L7_optical.filterBounds(Taluk).filterDate("2000-06-28","2000-12-15").filterMetadata('CLOUD_COVER','Less_than',10).mean().clip(Taluk)
vis_L72002432 = {'bands': ['SR_B4', 'SR_B3', 'SR_B2'], 'gamma': 3.62, 'min': 8410.0, 'max': 65535.0}
map.addLayer(Taluk_L7,vis_L72000543,'FCC_L72002543')
map.addLayer(Taluk_L7,vis_L72002432,'TCC_L72002432')
map
```

```python
def scale_landsat(image):
  optical=image.select(['SR_B1','SR_B2','SR_B3','SR_B4','SR_B5','SR_B7']).multiply(0.0000275).add(-0.2)
  return image.addBands(optical, overwrite=True)
```

```python
scaled_L7=scale_landsat(Taluk_L7)
```

```python
NIR_old=scaled_L7.select("SR_B4")
RED_old=scaled_L7.select("SR_B3")
```

```python
Taluk_NDVI2=NIR_old.subtract(RED_old).divide(NIR_old.add(RED_old)).rename("NDVI_old")
map.addLayer(Taluk_NDVI2,{},"Taluk_NDVIold")
map
```

```python
Landsat7_2001 = ee.ImageCollection("LANDSAT/LE07/C02/T1_L2")
L7_optical_2001 = Landsat7_2001.select('SR_B1','SR_B2','SR_B3','SR_B4','SR_B5','SR_B7')
Taluk_L7_2001 = L7_optical_2001.filterBounds(Taluk).filterDate("2001-06-28","2001-12-15").filterMetadata('CLOUD_COVER','Less_than',10).mean().clip(Taluk)
```

```python
scaled_L7_2001 = scale_landsat(Taluk_L7_2001)
```

```python
NIR = scaled_L7_2001.select("SR_B4")
RED = scaled_L7_2001.select("SR_B3")
```
